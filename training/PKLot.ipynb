```python
from datetime import datetime

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
run_name = f"train_{timestamp}"
print(run_name)
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
from ultralytics import YOLO
model = YOLO("yolov8n.pt")

model.train(
    data = "/content/drive/MyDrive/PKLot/data/data.yaml",
    epochs = 25,
    project='/content/drive/MyDrive/PKLot/yolo_runs',
    name = run_name,
    imgsz = 640,
    batch = 16,
    device = 0
)

```

```python
import json
import os
import shutil

results = model.val()

metrics_summary = {
    "precision": float(results.box.mp),
    "recall": float(results.box.mr),
    "mAP50": float(results.box.map50),
    "mAP50_95": float(results.box.map)
}

os.makedirs(f"/content/drive/MyDrive/PKLot/yolo_runs/{run_name}", exist_ok=True)

with open(f"/content/drive/MyDrive/PKLot/yolo_runs/{run_name}/metrics_summary.json", "w") as f:
    json.dump(metrics_summary, f, indent=4)
```
